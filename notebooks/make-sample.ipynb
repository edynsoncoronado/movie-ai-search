# Generar Muestra

Genera data/sample/omdb_sample.jsonl: 200 películas para tests y quickstart.

- 21 películas fijas, elegidas para consultas de prueba (perros, Star Wars,
  Blade Runner y títulos ambiguos: Titanic, Parasite, Coco).
- El resto, al azar entre películas conocidas (>= 20k votos), con semilla fija
  para que la muestra sea reproducible.

## Script

```python
SIZE = 200
MIN_VOTES = 20_000
```

```python
def votes(raw: dict) -> int:
    value = str(raw.get("imdbVotes", "0")).replace(",", "") # Convert to string and remove commas
    return int(value) if value.isdigit() else 0
```

```python
def main(src: str, dst: str) -> None:
    # Leer el JSONL con Polars.
    df = pl.read_ndjson(src, infer_schema_length=None)

    # Filtrar registros válidos mediante expresiones de Polars.
    valid_df = df.filter(
        (pl.col("raw").struct.field("Response") == "True")
        & pl.col("raw").struct.field("Plot").is_not_null()
        & ~pl.col("raw").struct.field("Plot").is_in(["", "N/A"])
    )

    valid = valid_df.to_dicts()

    # Seleccionar las películas ancla disponibles.
    picked = {
        r["raw"]["imdbID"]: r
        for r in valid
        if r["raw"]["imdbID"] in ANCHORS
    }

    # Construir el conjunto de películas candidatas.
    pool = [
        r
        for r in valid
        if votes(r["raw"]) >= MIN_VOTES
        and r["raw"]["imdbID"] not in picked
    ]

    # Selección aleatoria reproducible.
    random.seed(42)
    for record in random.sample(pool, SIZE - len(picked)):
        picked[record["raw"]["imdbID"]] = record

    # Reconstruir el DataFrame, ordenar y escribir el JSONL.
    result = pl.DataFrame(list(picked.values()), schema=df.schema).sort(
        pl.col("raw").struct.field("imdbID")
    )
    result.write_ndjson(dst)

    print(f"{result.height} películas -> {dst}")
```

```python
main(
    "../data/omdb_raw.jsonl",
    "../data/sample/omdb_sample.jsonl"
)
```
